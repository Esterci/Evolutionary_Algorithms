```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint
```

```python

def model_lotka_voltera(y, t, alpha, beta, gamma, sigma):
  L, A = y

  dLdt = alpha * L - beta * L * A

  dAdt = sigma * L * A - gamma * A

  return [dLdt, dAdt]

alpha = 1
beta = 0.1
gamma = 1.5
sigma = 0.075

l0 = 10
a0 = 5

t = np.linspace(0, 80, 2000)
P0 = (l0,a0)

mmargs = (alpha, beta, gamma, sigma)

Y = odeint(model_lotka_voltera, P0, t, args=(mmargs))
plt.plot(t, Y[:,0], label="A(t)")
plt.plot(t, Y[:,1], label="B(T)")
plt.legend(loc='upper right')
plt.show()
```
